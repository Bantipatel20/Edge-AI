```python
from torchvision import transforms
from PIL import Image
import os
import random

not_mixed_dir = "./Dataset/Not_Mixed"
augmented_dir = "./Dataset_Augmented/Not_Mixed"
os.makedirs(augmented_dir, exist_ok=True)

transform_aug = transforms.Compose([
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(20),
    transforms.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.2, hue=0.1)
])

all_images = [f for f in os.listdir(not_mixed_dir) if f.endswith(('.png', '.jpg', '.jpeg'))]

num_to_augment = 3000
for i in range(num_to_augment):
    img_name = random.choice(all_images)
    img = Image.open(os.path.join(not_mixed_dir, img_name)).convert('RGB')
    img_aug = transform_aug(img)
    img_aug.save(os.path.join(augmented_dir, f"aug_{i}_{img_name}"))
```

```python
import os

augmented_dir = "./Dataset_Augmented/Not_Mixed"
if not os.path.exists(augmented_dir):
    print(f"Folder does not exist: {augmented_dir}")
else:
    print(f"Contents of folder: {os.listdir(augmented_dir)}")

aug_images = [f for f in os.listdir(augmented_dir) if f.lower().endswith(('.png', '.jpg', '.jpeg'))]
print(f"Found {len(aug_images)} image files for renaming.")

start_index = 2451

for i, img_name in enumerate(aug_images, start=start_index):
    ext = os.path.splitext(img_name)[1]
    new_name = f"{i}_NON-MIXED2026-01-28-00-00-00{ext}"
    old_path = os.path.join(augmented_dir, img_name)
    new_path = os.path.join(augmented_dir, new_name)
    os.rename(old_path, new_path)

print(f"Renamed {len(aug_images)} images in-place, starting from {start_index}")
```

```python
import os
import shutil

# Paths
augmented_dir = "./Dataset_Augmented/Not_Mixed"  # renamed augmented images
not_mixed_dir = "./Dataset/Not_Mixed"            # main folder

# Make sure destination exists
os.makedirs(not_mixed_dir, exist_ok=True)

# List all augmented images
aug_images = [f for f in os.listdir(augmented_dir) if f.lower().endswith(('.png', '.jpg', '.jpeg'))]

print(f"Found {len(aug_images)} augmented images to copy.")

# Copy images into main Not_Mixed folder
for img_name in aug_images:
    src_path = os.path.join(augmented_dir, img_name)
    dst_path = os.path.join(not_mixed_dir, img_name)
    # Avoid overwriting if by chance a file with the same name exists
    if not os.path.exists(dst_path):
        shutil.copy(src_path, dst_path)
    else:
        print(f"Skipped {img_name}, already exists in destination.")

print(f"Copied {len(aug_images)} images to {not_mixed_dir}")
```
